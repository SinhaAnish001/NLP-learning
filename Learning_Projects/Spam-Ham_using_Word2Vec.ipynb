```python
import pandas as pd
df = pd.read_csv('../data/modified/Spam_cleaned.csv')
```

```python
import nltk
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
stop_words = set(stopwords.words('english'))
```

```python
corpus=[]
for i in range(len(df)):
    review=re.sub('[^a-zA-z]',' ', df['Text'][i])
    review=review.lower()
    review=review.split()

    review = [lemmatizer.lemmatize(word) for word in review if word not in stop_words]

    review=' '.join(review)
    corpus.append(review)
```

```python
[[i,j,k] for i,j,k in zip(list(map(len, corpus)),corpus,df['Text']) if i<1]
```

```python
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
```

```python
words=[]
for sent in corpus:
    sent_token= sent_tokenize(sent)
    for sent in sent_token:
        words.append(simple_preprocess(sent))
```

```python
# For every word in the document, get its Word2Vec vector only if the word exists in the vocabulary."
def avg_word2vec(doc):
    vectors = [model.wv[word] for word in doc if word in model.wv.index_to_key]

    if len(vectors) == 0:
        return np.zeros(model.vector_size)

    return np.mean(vectors, axis=0)
```
