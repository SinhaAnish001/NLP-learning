## SPAM Classifier using BOW 

```python
import pandas as pd
df = pd.read_csv('../data/modified/Spam_cleaned.csv')
```

```python
df.head()
```

```python
## Data cleaning and preprocessing
import nltk
import re
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
stop_words = set(stopwords.words('english'))
```

```python
corpus=[]
for i in range(len(df)):
    review=re.sub('[^a-zA-z]',' ', df['Text'][i])
    review=review.lower()
    review=review.split()

    review = [ps.stem(word) for word in review if word not in stop_words]

    review=' '.join(review)
    corpus.append(review)
```

```python
corpus
```

```python
## Output features
y=pd.get_dummies(df['Target'])
y=y.iloc[:,0].values
```

```python
## Train test split
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(corpus,y,test_size=0.2)
```

## Create Bag of Words

```python
## create BOW model
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=2500,binary=True,ngram_range=(1,2)) 
```

```python
X_train = cv.fit_transform(X_train).toarray()
X_test = cv.transform(X_test).toarray()
```

```python
import numpy as np
np.set_printoptions(edgeitems=30,linewidth=100000,
    formatter = dict(float=lambda x: "%.3g" % x))
```

```python
X_train
```

```python
cv.vocabulary_
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
spam_detection_model = MultinomialNB().fit(X_train,y_train)
```

```python
spam_detection_model
```

```python
y_pred=spam_detection_model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```
