## Tokenization Basics

```python
corpus = """Hello Welcome , to Anish Sinha's NLP practice.
Please do complete the entire course! to become expert in NLP.
"""
```

```python
print(corpus)
```

```python
# sentence --> paragrapgh

from nltk.tokenize import sent_tokenize
document = sent_tokenize(corpus)
```

```python
type(document)
```

```python
for sentence in document:
    print(sentence)
```

```python
# paragrapgh --> words
# sentence --> words

from nltk import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in document:
    print (word_tokenize(sentence))
```

```python
# to seperate ' & s from sinha's
from nltk import wordpunct_tokenize
wordpunct_tokenize(corpus)
```
