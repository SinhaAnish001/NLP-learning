## STEMMING : process of reducing words to its word stem
- eating ---> eat
- [going,gone] ---> go

### 1. PorterStemmer

```python
from nltk import PorterStemmer
```

```python
words = ["running", "jumps", "fairly","sportingly","going", "history"]
```

```python
stemming= PorterStemmer()
```

```python
for word in words:
    print(word+"--->"+stemming.stem(word))
```

## Disadvantage : Entire meaning of words can change

### 2. RegexpStemmer : remove the suffix you want from regular expression

```python
from nltk import RegexpStemmer
word2 = ["running","advisable","jumping","miserable","eats"]
```

```python
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
for word in word2:
    print(word+"--->"+reg_stemmer.stem(word))
```

### 3. SnowballStemmer : same as PorterStemmer but better for some words

```python
from nltk import SnowballStemmer
ss = SnowballStemmer("english")
```

```python
for word in words:
    print(word+"--->"+ss.stem(word))
```

```python
stemming.stem("fairly"), stemming.stem("sportingly")
```

```python
ss.stem("fairly"), ss.stem("sportingly")
```
