## WordNetLemmatizer : 
### Tech. like stemming. The output we will get after lemmatization is called 'lemma' , which is a WORD ROOT rather than root stem

```python
from nltk import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

```python
import nltk
nltk.download('wordnet')
```

```python
''' param pos : The Part Of Speech tag. 
Valid options are "n" for nouns,
"v" for verbs, 
"a" for adjectives, 
"r" for adverbs and
"s" for satellite adjectives.
'''
lemmatizer.lemmatize("going", pos='v')
```

```python
words = ["running", "jumps", "fairly","sportingly","going", "history"]
for word in words:
    print(word+"--->"+lemmatizer.lemmatize(word,pos='v'))
```
